## **scatter Plot**

```python
x=np.array([1,2,3,4])
y=np.array([1,2,3,4])

plt.scatter(x,y, c='r',marker = "*", s=100 )  # c for colour , marker for point shape , s for size of the marker point ie * star size 
plt.show();
```

```python
# another exampler for scatter

x=np.array([1,2,3,4])
y=np.array([1,2,3,4])
color=["red","yellow","green","black"]

plt.scatter(x,y, c= color ,marker = "*", s=100 )  # c for colour , marker for point shape , s for size of the marker point ie * star size 
plt.show();
```

## **Line graph**

```python
#Example 1
x=np.array([1,2,3,4])
y=np.array([1,2,3,4])
plt.plot(x,y)
plt.show();
```

```python
# Another example 
x=np.array([1,2,3,4])
y=np.array([1,2,3,4])
plt.plot(x,y,c='g',linestyle="dashed") # c fot green colour , linestyle="dashed" in graph
plt.scatter(x,y,c = color,marker = "*",s=100) # scatter is using for point
plt.show();
```

## **Bar chart**

```python
year=np.array([1990,2000,2010,2020])
lang=np.array(["c","c++","java","python"])
plt.bar(year,lang)
plt.show();
```

```python
year = ["1990","2000","2010","2020","2030"]
lang = ["None","Java","C","C++","Python"]
c = ['Green','darkorange','deeppink','violet',"red"]
plt.bar(year,lang,color = c,align="edge") # in bar graph we should give ful name color not single character " c" bar graph cant understand c for color
                                           #align="edge" start from barchart edge
plt.show();
```

```python
# frquency distribution means histrogram chart
np.random.seed(2)
a = np.random.normal(80,2,100) # 80 is the peak value , 2 is the difference , total 100 value
plt.hist(a,color='deeppink',bins=15,edgecolor="black") #bins means barchart count 
plt.show()
```

## **Pie Chart**

```python
lang =["python","java","c++","c"]
votes=[90,75,50,40]
exp = [0.1,0,0,0] #cutting pieces value for pie chart 
c = ['Green','darkorange','deeppink','violet']
plt.pie(votes,labels=lang,colors=c,explode=exp) # explode for cutting pieces  
plt.show();
```

```python
#Example for title ,xlaabel and ylabel in plot (line chart)

years=["2000","2010","2020","2030"]
income = [200,400,600,800]
plt.title("Daily Income of Manjunathan")
plt.xlabel("Years")
plt.ylabel("Income")
plt.plot(years,income,c="g",marker="*")
plt.show()
```

## **Matplotlib session 2**

```python
stockA=[1,2,3,4,5]
stockB=[3,4,5,3,2]
stockC=[6,4,2,1]
plt.title("stock market")


plt.plot(stockA ,label="stock1" ,marker="*") # label="stock1" is used in the legend. 
plt.plot(stockB, label="stock2", marker="p")
plt.plot(stockC,label="stock3",marker="o")
plt.legend()   # display the labels in the box 
   
plt.show();
```

## **1.method (pyplot api)**

```python
fig=plt.figure() # fig is the default 
ax=fig.add_subplot()   # ax is the default name for add_subplot

ax.plot([1,2,3,5],[-1,2,3,4],c="g")
plt.show();
```

## **2.method (object oriented api)** *(real time we are using method 2 only)*

```python
fig,ax=plt.subplots() # we are combining the fig and ax in oneline as subplots
plt.plot([1,2,3,5],[-1,2,3,4],c="g")
plt.show();
```

```python
# creating dictionary and plot 
carmileage={
    "BMW": 7 ,
    "FORD": 10 ,
    "INNOVA": 15
}
fix,ax=plt.subplots()
ax.bar(carmileage.keys(),carmileage.values());
plt.show();
    
```

```python
# creating dictionary and plot and also changing the bar vertivcal to horizontal

car_milage ={
"BMW":7,
"Ford":10,
"Toyota":15
}
fig, ax = plt.subplots()
ax.barh(car_milage.values(),car_milage.keys());
plt.show()
```

```python
#changing graph values in X to y axis and y to x axis we need to use  list 

carmileage={
    "BMW": 7 ,
    "FORD": 10 ,
    "INNOVA": 15
}
fix,ax=plt.subplots()
ax.barh(list(carmileage.keys()),(carmileage.values()),color=["b","pink","orange"]); # barh is showing in horizontal 
plt.show();
```

## **subplots comparing (age vs chol),(age vs thalach )**

```python

fig,ax=plt.subplots(nrows=2,ncols=1,figsize=(12,8))
#subplot=1
scatter = ax[0].scatter(df["age"],df["chol"],c=df["target"],cmap="winter") #cmap is the colour chat check in google

#customise the plot
ax[0].set(title ='Age vs Cholestrol',xlabel="Age",ylabel="Cholestrol")
ax[0].legend(*scatter.legend_elements(),title="Target")
#add mean line finfing mean in the chol
ax[0].axhline(df["chol"].mean(),color="r",linestyle="dashed") 


scatter=ax[1].scatter(df["age"],df["thalach"],c=df["target"],cmap="viridis")
#customise the plot
ax[1].set(title='Age vs Thalach',xlabel="Age",ylabel="Thalach")
ax[1].legend(*scatter.legend_elements(),title="Target")
ax[1].axhline(df["thalach"].mean(),color="r",linestyle="dashed")
plt.subplots_adjust(hspace=0.4)  # plt.subplots_adjust(hspace=0.4) is the distance between 2 graph in image we are using hspace=0.4
plt.show();
```

## **save as image this plot**

```python



fig, ax = plt.subplots(figsize=(10, 6))
# Plot Data
scatter = ax.scatter(df['age'], df['chol'], c=df['target'],cmap="winter")
# Customize the plot
ax.set(title="Heart Disease and Cholesterol Levels", xlabel="Age",ylabel="Cholesterol")
ax.legend(*scatter.legend_elements(), title='Target')
# Add a meanline
ax.axhline(df['chol'].mean(), linestyle='dashed');

plt.savefig("Probability_of_heart_attack.png") # saving the image after save we cant able to edit we can see this image in left side
```

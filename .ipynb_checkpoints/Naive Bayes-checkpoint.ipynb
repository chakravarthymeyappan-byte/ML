## **Naive Bayes**

```python
import pandas as pd
sentimental_data=pd.read_csv("C:\\Users\\chakravarthy\\OneDrive\\ドキュメント\\csv_files\\Naive_Bayes_Sentiment_Analysis_Dataset_30_Rows_Sentiment_Dataset.csv",encoding ="unicode_escape")
sentimental_data
```

```python
sentimental_data.info()
```

```python
sentimental_data.isna().sum()
```

```python
sentimental_data.columns=[cols.lower() for cols in sentimental_data]
sentimental_data
```

```python
input_data=sentimental_data['review_text']
output_data=sentimental_data['sentiment']
```

```python
input_data
```

```python
output_data
```

```python
print(input_data.shape)
print(output_data.shape)
```

```python
from sklearn.model_selection import train_test_split
input_data_train,input_data_test,output_data_train,output_data_test=train_test_split(input_data,output_data,test_size=0.2)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
#TfidfVectorizer converts text data into numerical values so that a Machine Learning model can understand the text.
from sklearn.naive_bayes import MultinomialNB
#MultinomialNB is a Machine Learning classification algorithm based on Bayes' theorem.
from sklearn.pipeline import make_pipeline
#make_pipeline() combines multiple Machine Learning steps into one pipeline.
```

```python
model=make_pipeline(TfidfVectorizer(),MultinomialNB())
```

```python
model.fit(input_data_train,output_data_train)
```

```python
predicted_sentiment=model.predict(input_data_test)
```

```python
predicted_sentiment
```

```python
print(input_data_train.shape)
print(input_data_test.shape)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(output_data_test,predicted_sentiment)
```

```python
pd.DataFrame(confusion_matrix(output_data_test,predicted_sentiment),columns=["predicted_Negative","predicted_Neutral","preticted_Positive"],index=["Actual_Negative","Actual_Neutral","Actual_positive"])
```

```python
from sklearn.metrics import accuracy_score
acccuracy_info=accuracy_score(output_data_test,predicted_sentiment)
acccuracy_info
```

```python
import pandas as pd
tel_data=pd.read_csv("C:\\Users\\chakravarthy\\OneDrive\\ドキュメント\\csv_files\\telescope_data.csv")
tel_data
```

```python
tel_data.columns=[cols.lower() for cols in tel_data]
```

```python
import matplotlib.pyplot as plt
for label in (tel_data.columns):
    plt.hist(tel_data[tel_data['class']=="g"][label],color="green",label='gamma',alpha=0.5,density=True) # alpha is the clour tnickness bright or lighter(Makes the histogram transparent) and density=True (Displays normalized density)
    plt.hist(tel_data[tel_data['class']=='h'][label],color="yellow",label='hadron',alpha=0.5,density=True)
    plt.title(label)
    plt.ylabel("probability")
    plt.xlabel(label)
    plt.legend()
    plt.show();
    
```

```python
from sklearn.model_selection import train_test_split
ds_train,ds_test= train_test_split(tel_data,test_size=0.2)
```

```python
print(len(ds_train))
print(len(ds_test))
```

```python
print(len(ds_train[ds_train['class']=='g']))
print(len(ds_train[ds_train['class']=='h']))
```

```python
!pip install imbalanced-learn
```

```python
from imblearn.over_sampling import RandomOverSampler
r_ram = RandomOverSampler() #r_ram is the object name for class RandomOverSampler

#splitting input and output data

input_ds_train=ds_train.drop(columns=['class'])
output_ds_train=ds_train['class']
```

```python
print(input_ds_train)
```

```python
print(output_ds_train)
```

```python
#normalising data ie float percentage use import StandardScaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
input_ds_train= scaler.fit_transform(input_ds_train)
print(input_ds_train)
```

```python
input_ds_train,output_ds_train = r_ram.fit_resample(input_ds_train,output_ds_train)
print(len(output_ds_train))
print(sum(output_ds_train == 'g'))
print(sum(output_ds_train == 'h'))
```

```python
input_ds_test = ds_test.drop(columns = ['class'])
output_ds_test=ds_test['class']
input_ds_test= scaler.fit_transform(input_ds_test)
```

```python
input_ds_test
```

```python
print(len(input_ds_test))
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Create KNN model
knn_model = KNeighborsClassifier(n_neighbors=2)#Knn_model – Object name and The model considers 2 nearest neighbours.

# Train the model
knn_model.fit(input_ds_train, output_ds_train)
```

```python
output_pred=knn_model.predict(input_ds_test)
```

```python
output_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_info =accuracy_score(output_ds_test,output_pred)
accuracy_info
```

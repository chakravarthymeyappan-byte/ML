```python
import os
import csv
import pandas as pd
import matplotlib.pyplot as plt

path =r"C:\Users\chakravarthy\OneDrive\Desktop"
filename = "shipping.txt"
```

```python
# step 1 Read the log file 
status=[]
issue=[]
with open(os.path.join(path ,filename),"r") as f:
    for each in f:
        mylist=each.split(" ")
        status.append(mylist[6])
        issue.append(mylist[7])
        
#print(status)
#print(issue)
```

```python
# frequency 
def frequency(mylist):  # creating function
    result = {}         # creating dictionary for storing data

    for each in mylist:
        if each not in result:
            result[each] = 1
        else:
            result[each] += 1    # increment count by 1

    return result

#print(frequency(status))
#print(frequency(issue))
```

```python
# creating csv file 

def status_csv1(data):
    with open(os.path.join(path,"status.csv"),'w',newline = '') as n:
        writer = csv.writer(n)
        writer.writerow(['Sno','Error_status','frequency'])
        Sno=1
        for each in data:
            writer.writerow([Sno,each,data[each]])
            Sno += 1
            
        
                        
        
```

```python
def  issue_csv2(data):
    with open(os.path.join(path,"issue.csv"),'w',newline = '') as n:
        writer = csv.writer(n)
        writer.writerow(['Sno','Error_issue','frequency'])
        Sno=1
        for each in data:
            writer.writerow([Sno,each,data[each]])
            Sno += 1

        
```

```python
obj = frequency(status)
obj1 = frequency(issue)   
        
status_csv1(obj)
issue_csv2(obj1)
print("Unique status and their counts:")
print(obj)
print(" issue code and  counts:")
print(obj1)
print('CSV file created sucessfully')
```

```python
#importing our csv file to jupyter lab
file1=pd.read_csv("C:\\Users\\chakravarthy\\OneDrive\\Desktop\\status.csv")
file1
```

```python
fig, ax = plt.subplots()
ax.bar(file1['Error_status'].astype(str),file1['frequency'], color='blue',width=0.6)
ax.set_xlabel('Error Status')
ax.set_ylabel('Frequency')
ax.set_title('Error Status Frequency')
# Rotate x-axis labels
plt.xticks(rotation=20, ha='right') #xticks is the rotation of the wording 

# Adjust layout automatically
plt.tight_layout()

plt.show()
```

```python

# splitting input and output data
input_data=titanic_data.drop(columns=["name","survived","cabin","embarked","ticket"])
```

```python
input_data
```

```python
output_data=titanic_data["survived"]
```

```python
output_data
```

```python
# StandardScaler is converting into normalised form ie convert into percentage
from sklearn.preprocessing import StandardScaler
scalar=StandardScaler() # scalar is the object name 
input_data=scalar.fit_transform(input_data)
print(input_data)
```

```python
from sklearn.model_selection import train_test_split
input_data_train,input_data_test,output_data_train,output_data_test=train_test_split(input_data,output_data,test_size=0.2)
```

```python
input_data_train.shape
```

```python
input_data_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
model= LogisticRegression()

model.fit(input_data_train,output_data_train)
```

```python
predicted_survived=model.predict(input_data_test)
```

```python
predicted_survived
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(output_data_test,predicted_survived)
```

```python
pd.DataFrame(confusion_matrix(output_data_test,predicted_survived),columns=["predicted yes","predicted No"], index=["Actual yes", "Actual No"])
```

```python
from sklearn.metrics import accuracy_score
accuracy_info=accuracy_score(output_data_test,predicted_survived)
```

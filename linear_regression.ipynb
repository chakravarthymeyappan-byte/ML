```python
#spliting the input and output data

input_data=car_info.drop(columns=["car_name","selling_price"])

output_data=car_info["selling_price"]

```

```python
input_data = pd.get_dummies(input_data, dtype=int)
print(input_data.head())
```

```python
input_data.shape
```

```python
#StandardScaler convert all the values input data into percentage
from  sklearn.preprocessing import StandardScaler #class name : StandardScaler 

scalar= StandardScaler()  # scalar is the object name for class StandardScaler 

input_data=scalar.fit_transform(input_data)
print(input_data)
```

```python

from sklearn.model_selection import train_test_split
input_data_train,input_data_test,output_data_train,output_data_test= train_test_split(input_data,output_data,test_size=0.2)
```

```python
print(input_data_train.shape) # checking how many data taken for train to machine
```

```python
print(input_data_test.shape) # checking how many data taken for test
```

```python
from sklearn.linear_model import LinearRegression
from sklearn import metrics
model=LinearRegression()
model.fit(input_data_train,output_data_train)
```

```python
predicted_selling_price=model.predict(input_data_test)
```

```python
predicted_selling_price
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
```

```python
print("mean absolute error",(metrics.mean_absolute_error(predicted_selling_price,output_data_test)))
print("mean square error",(metrics.mean_squared_error(predicted_selling_price,output_data_test)))
```
